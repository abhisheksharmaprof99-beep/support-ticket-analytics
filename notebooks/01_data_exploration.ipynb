```python
df["First Response Time"] = pd.to_datetime(df["First Response Time"])
df["Time to Resolution"] = pd.to_datetime(df["Time to Resolution"])

closed = df[df["Ticket Status"] == "Closed"].copy()
closed["gap_hours"] = (closed["Time to Resolution"] - closed["First Response Time"]).dt.total_seconds() / 3600

print("Closed tickets:", len(closed))
print("Resolved BEFORE first response:", (closed["gap_hours"] < 0).sum())
print(closed["gap_hours"].describe())
print("Response date range:", df["First Response Time"].min(), "to", df["First Response Time"].max())
```
